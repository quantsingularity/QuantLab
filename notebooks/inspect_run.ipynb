## Inspect a QuantLab run

Loads the artefacts produced by `python -m quantlab.run --objective "..."` the equity curve, trade log, metrics, and reflective critiques and renders them for quick visual sanity-checking.

Run the PoC first if you haven't:
```bash
python -m quantlab.run --objective "Develop a momentum strategy for the NASDAQ 100" --out ./runs/momentum_nasdaq
```

````python
import json
from pathlib import Path

import pandas as pd

# Point this at any completed run directory under ./runs/<run_id>/
RUN_DIR = Path("../runs/momentum_nasdaq")
run_id = next(p.name for p in RUN_DIR.iterdir() if p.is_dir())
run_dir = RUN_DIR / run_id
print(f"Inspecting run: {run_id}")
run_dir
````

````python
metrics = json.loads((run_dir / "metrics.json").read_text())
metrics
````

````python
equity = pd.read_parquet(run_dir / "equity_curve.parquet")
equity.plot(title="Equity curve (net of costs)", figsize=(9, 4), grid=True)
````

````python
trades = pd.read_parquet(run_dir / "trades.parquet")
trades.head()
````

````python
reflections_path = run_dir / "reflections.jsonl"
reflections = [
    json.loads(line) for line in reflections_path.read_text().splitlines() if line
]
pd.DataFrame(reflections)
````
